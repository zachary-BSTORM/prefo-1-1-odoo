- La fonction reçoit une matrice carrée (autant de lignes que de colonnes) et renvoie la somme des
éléments de ses deux diagonales :
    - la diagonale principale, d'en haut à gauche à en bas à droite ;
    - la diagonale secondaire, d'en haut à droite à en bas à gauche.
    - Une case qui appartient aux deux diagonales n'est comptée qu'une seule fois.

```python
def somme_diagonales(matrice: list[list[int]]) -> int:
    n = len(matrice)

    total = 0

    for i in range(n):
        total += matrice[i][i]

        total += matrice[i][n - 1 - i]

    if n % 2 == 1:
        millieu = n // 2

        total -= matrice[millieu][millieu]

    return total


print(somme_diagonales([[1, 2, 3], [4, 5, 6], [7, 8, 9]]))  # 25
```

- La fonction reçoit un entier strictement positif n et renvoie True si n est un nombre heureux, False sinon.
- Pour savoir si un nombre est heureux, on répète l'opération suivante :
    - on remplace le nombre par la somme des carrés de ses chiffres (par exemple 19 devient 1² + 9² = 82) ;
    - on recommence avec le nouveau nombre.
- Si on finit par tomber sur 1, le nombre est heureux. Sinon, les valeurs se mettent à tourner en rond indéfiniment sans jamais passer par 1 : le nombre n'est pas heureux.

```python
def est_heureux(n: int) -> bool:
    deja_vus = set()
    
    while n != 1:
        
        if n in deja_vus:
            return False
        
        deja_vus.add(n)
        
        somme = 0
        while n > 0:
            chiffre = n % 10
            somme += chiffre * chiffre
            n = n // 10
        n = somme
        
    return True
    
print(est_heureux(19))
print(est_heureux(2))
```

- La fonction reçoit une liste non vide d'entiers positifs ou nuls. 
- On démarre sur la première case (indice 0).
- La valeur de chaque case indique la longueur maximale du saut possible depuis cette case : 
    - depuis une case qui vaut 3, on peut avancer de 1, 2 ou 3 cases (ou ne pas bouger). 
    - Depuis une case qui vaut 0, on ne peut plus avancer.

- La fonction renvoie True s'il est possible d'atteindre la dernière case
- Sinon False

```python
def peut_atteindre_fin(sauts: list[int]) -> bool:
    portee = 0

    for i in range(len(sauts)):
        if i > portee:
            return False

        portee = max(portee, i + sauts[i])
    return True


# print(peut_atteindre_fin([2, 3, 1, 1, 4]))
# print(peut_atteindre_fin([3, 2, 1, 0, 4]))


def jeu_du_saut(sauts: list[int]):
    for i in range(len(sauts)):
        if i == len(sauts) - 1:
            return True
        if sauts[i] == 0:
            # On inite un index pour aller en arrière
            j = i - 1
            while j >= 0 and j + sauts[j] <= i:
                j -= 1
            if j < 0:
                return False
    return True


print(jeu_du_saut([2, 0, 0]))
```

- Des maisons sont alignées dans une rue. La liste maisons donne la somme d'argent disponible dans
chacune, dans l'ordre de la rue (tous les montants sont positifs ou nuls).
- Un cambrioleur veut emporter le maximum d'argent, mais il ne peut pas cambrioler deux maisons
voisines, sinon l'alarme se déclenche.
- La fonction renvoie le montant maximum qu'il peut emporter. Si la liste est vide, elle renvoie 0.

```python
def cambrioleur(maisons: list[int]) -> int:
    dernier = 0
    avant_dernier = 0

    for m in maisons:
        prend = avant_dernier + m

        prend_pas = dernier

        meilleur = max(prend, prend_pas)

        avant_dernier = dernier
        dernier = meilleur

    return dernier

print(cambrioleur([2, 7, 9, 3, 1]))
print(cambrioleur([2, 1, 1, 2]))
```

```python
def cambrioleur_recursif(maisons: list[int]) -> int:
    if len(maisons) == 0:
        return 0

    if len(maisons) == 1:
        return maisons[0]
    if len(maisons) == 2:
        return max(maisons[0], maisons[1])

    return max(
        maisons[0] + cambrioleur_recursif(maisons[2:]), maisons[1] + cambrioleur_recursif(maisons[3:])
    )


print(cambrioleur_recursif([2, 7, 9, 3, 1]))
print(cambrioleur_recursif([2, 1, 1, 2]))
```

- La fonction reçoit une matrice carrée n × n (une image dont chaque case est un pixel) 
    - renvoie la matrice tournée de 90° dans le sens des aiguilles d'une montre.
- La fonction peut modifier la matrice reçue directement (sur place) et la renvoyer.

```python
def rotation_90(matrice: list[list[int]]) -> list[list[int]]:
    n = len(matrice)

    new = []

    for _ in range(n):
        new.append([0] * n)

    for i in range(n):
        for j in range(n):
            new[j][n - 1 - i] = matrice[i][j]
    return new

def rotation_90_2(matrice: list[list[int]]) -> list[list[int]]:
    n = len(matrice)
    
    for i in range(n):
        for j in range(i + 1 , n):
            matrice[i][j] , matrice[j][i] =  matrice[j][i],matrice[i][j]

    for l in matrice:
        l.reverse()
        
    return matrice

def rotation_90_3(matrice: list[list[int]]) -> list[list[int]]:
    return [ list(r) for r in zip(*matrice[::-1])]

result = rotation_90_3([[1, 2, 3], [4, 5, 6], [7, 8, 9]])  # [[7,4,1],[8,5,2],[9,6,3]]

for v in result:
    print(v)
```

- La fonction reçoit deux mots a et b et renvoie True si ce sont
des anagrammes, False sinon.
- Deux mots sont des anagrammes s'ils contiennent exactement les
mêmes lettres, chacune le même nombre de fois, dans un ordre
quelconque.
- Les majuscules et les minuscules sont des lettres différentes.

```python
def sont_anagrammes(a: str, b: str) -> bool:
    return sorted(a) == sorted(b)


def sont_anagrammes(a: str, b: str) -> bool:
    if len(a) != len(b):
        return False

    compte = {}

    for c in a:
        compte[c] = compte.get(c, 0) + 1

    for c in b:
        if compte.get(c, 0) == 0:
            return False
        compte[c] -= 1

    return True


print(sont_anagrammes("chien", "niche"))  # True
print(sont_anagrammes("aab", "abb"))  # False (pas le même nombre de a)
print(sont_anagrammes("chat", "chien"))  # False (longueurs différentes)
print(sont_anagrammes("", ""))  # True (deux mots vides)
print(sont_anagrammes("listen", "silent"))  # True
print(sont_anagrammes("Rome", "more"))  # False (R majuscule ≠ r)
print(sont_anagrammes("aabb", "abab"))  # True
```

- La fonction reçoit deux mots. Le mot a sert de stock de lettres :
chaque lettre de a ne peut être utilisée qu'une seule fois.
- La fonction renvoie True si on peut former le mot b avec les
lettres de a , et False sinon. Il peut rester des lettres inutilisées
dans a .
- C'est le même principe que les anagrammes, mais a peut contenir
des lettres en plus.

```python
def peut_construire(a: str, b: str) -> bool:
    stock = {}
    for c in a:
        stock[c] = stock.get(c, 0) + 1

    for c in b:
        if stock.get(c, 0) == 0:
            return False

        stock[c] -= 1

    return True

print(peut_construire("aabbc", "abc"))  # True
print(peut_construire("abc", "aab"))  # False (un seul a)
print(peut_construire("abc", ""))  # True (mot vide)
print(peut_construire("", "a"))  # False (stock vide)
print(peut_construire("magazine", "zen"))  # True
print(peut_construire("aa", "aaa"))  # False (un a de trop)
print(peut_construire("aabbc", "abcc"))  # False (un seul c)
```

- Le mot b a été fabriqué en mélangeant les lettres de a , puis en
ajoutant une seule lettre à une position quelconque.
- La fonction renvoie cette lettre ajoutée.
- Attention : la lettre ajoutée peut déjà exister dans a .

```python
def lettre_en_trop(a: str, b: str) -> str:
    stock = {}
    for c in a:
        stock[c] = stock.get(c, 0) + 1

    for c in b:
        if stock.get(c, 0) == 0:
            return c

        stock[c] -= 1

    return ""


print(lettre_en_trop("abcd", "dbcae"))  # e
print(lettre_en_trop("aab", "abaa"))  # a (la lettre existait déjà)
print(lettre_en_trop("", "y"))  # y (a est vide)
print(lettre_en_trop("xyz", "zxyx"))  # x
print(lettre_en_trop("abc", "cbad"))  # d
print(lettre_en_trop("zz", "zzz"))  # z (une seule lettre différente)
```

- La fonction reçoit une liste d'entiers (positifs, négatifs ou nuls) et
renvoie la liste des indices des éléments strictement supérieurs à
la somme de tous les éléments situés à leur gauche.
- Pour le premier élément, il n'y a rien à gauche : la somme vaut 0.
- Les indices sont renvoyés dans l'ordre croissant. Si la liste est vide,
la fonction renvoie une liste vide.

```python
def superieurs_a_gauche(nombres: list[int]) -> list[int]:
    resultat = []

    somme = 0

    for i, v in enumerate(nombres):
        if v > somme:
            resultat.append(i)

        somme += v

    return resultat


print(superieurs_a_gauche([1, 2, 4, 3, 12]))  # [0, 1, 2, 4]
print(superieurs_a_gauche([3, 1, 1, 8]))  # [0, 3]
print(superieurs_a_gauche([]))  # [] (liste vide)
print(superieurs_a_gauche([5]))  # [0] (seul : 5 > 0)
print(superieurs_a_gauche([0, 0]))  # [] (0 > 0 est faux)
print(superieurs_a_gauche([-1, 5]))  # [1] (avec un négatif)
print(superieurs_a_gauche([1, 1, 1]))  # [0]

print(superieurs_a_gauche([5, -10, 3]))  # [0, 2]
print(superieurs_a_gauche([1, 2, 3, 6]))  # [0, 1]
```

- La fonction reçoit une liste d'entiers et renvoie le nombre
d'éléments dont l'élément précédent est exactement plus grand de
1, c'est-à-dire tels que nombres[i - 1] == nombres[i] + 1 .
- Le premier élément n'a pas de précédent : il n'est jamais compté.
Pour une liste vide ou d'un seul élément, la fonction renvoie 0.

```python
def compter_descentes(nombres: list[int]) -> int:
    total = 0
    
    for i in range(1 , len(nombres)):
        if nombres[i -1] == nombres[i] + 1:
            total += 1
            
    return total
    
    
    
print(compter_descentes([5, 4, 3, 7, 6, 2])) # 3
print(compter_descentes([1, 2, 3])) # 0 (que des montées)
print(compter_descentes([])) # 0 (liste vide)
print(compter_descentes([9])) # 0 (un seul élément)
print(compter_descentes([3, 2, 2, 1])) # 2
print(compter_descentes([0, -1, -2])) # 2 (avec des négatifs)
print(compter_descentes([3, 7, 4])) # 0 (le piège de nombres[-1])
```

La fonction reçoit un mot non vide et renvoie True s'il peut s'écrire
comme un morceau répété au moins deux fois, collé bout à bout, et
False sinon.
Par exemple, « abcabcabc » est le morceau « abc » répété 3 fois.

```python
def est_repetition(mot: str) -> bool:
    n = len(mot)
    
    for l in range(1 , n // 2 + 1):
        
        if n % l == 0 :
            part = mot[:l]
            
            if part * (n // l) == mot:
                return True
        
    return False

print(est_repetition("abab")) # True
print(est_repetition("abcabcabc")) # True
print(est_repetition("aaaa")) # True ("a" × 4)
print(est_repetition("aba")) # False
print(est_repetition("a")) # False (une seule lettre)
print(est_repetition("abac")) # False
print(est_repetition("abaababaab")) # True ("abaab" × 2)


# "abcabcabc"
```

La fonction reçoit une matrice de lettres minuscules et cherche un
carré 2×2 dont les quatre lettres sont des voyelles ( a , e , i , o ,
u ).
Elle renvoie la position [i, j] du coin en haut à gauche de ce
carré. S'il y en a plusieurs, on garde le plus haut, puis le plus à
gauche.
S'il n'y en a aucun, elle renvoie [-1, -1] .

```python
def carre_voyelles(matrice: list[list[str]]) -> list[int]:
    voyelles = {"a" , "e" , "i" , "o" , "u"}
    
    nb_lignes = len(matrice)
    
    nb_colonnes = len(matrice[0]) if nb_lignes > 0 else 0
    
    for i in range(nb_lignes - 1):
        for j in range(nb_colonnes - 1):
            
            # (i,j) , (i , j + 1) , (i + 1 , j ) , (i + 1 , j + 1) 
            if matrice[i][j] in voyelles and matrice[i+1][j] in voyelles and matrice[i][j + 1] in voyelles and matrice[i+ 1][j + 1] in voyelles :
                return [i,j]
    return [ -1 , -1]
    
    
print(carre_voyelles([["x","a","e"], ["a","i","o"], ["u","e","k"]])) # [0, 1]
print(carre_voyelles([["a","b"], ["c","e"]])) # [-1, -1]
print(carre_voyelles([["a","e"]])) # [-1, -1] (une seule ligne)
print(carre_voyelles([])) # [-1, -1] (matrice vide)
print(carre_voyelles([["a","e","i"], ["o","u","a"]])) # [0, 0]
print(carre_voyelles([["k","k","k"], ["k","a","e"], ["k","i","o"]])) # [1, 1] (en bas à droite)
print(carre_voyelles([["a","b","a","e"], ["e","i","o","u"]])) # [0, 2] (pas le premier coin)
```

La fonction reçoit une liste d'entiers strictement positifs et un
entier cible strictement positif. Elle cherche une suite
d'éléments contigus (pris d'un seul tenant, sans en sauter) dont la
somme vaut exactement cible .
Elle renvoie [debut, fin] , les indices du premier et du dernier
élément de cette fenêtre. S'il y en a plusieurs, on renvoie la plus à
gauche. S'il n'y en a aucune, elle renvoie [-1, -1] .

```python
def fenetre_cible(nombres: list[int], cible: int) -> list[int]:
    debut = 0
    somme = 0
    
    for fin in range(len(nombres)):
        somme += nombres[fin]
        
        while somme > cible and debut <= fin:
            somme -= nombres[debut]
            debut += 1
    
        if somme == cible:
            return [debut,fin]
        
    return [-1,-1]
    
print(fenetre_cible([1, 3, 2, 5, 1, 1], 8)) # [2, 4]
print(fenetre_cible([4, 2, 6], 6)) # [0, 1]
print(fenetre_cible([1, 2, 3], 7)) # [-1, -1] (aucune fenêtre)
print(fenetre_cible([5], 5)) # [0, 0] (un seul élément)
print(fenetre_cible([1, 1, 1, 1], 3)) # [0, 2]
print(fenetre_cible([10, 1, 2, 3], 6)) # [1, 3] (le 10 est retiré)
print(fenetre_cible([2, 4], 1)) # [-1, -1] (cible plus petite que tout)
```
